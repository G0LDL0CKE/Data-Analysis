```python
# Read the csv file sep='|'
df = pd.read_csv("./2024/bitpanda-trades-2025-03-08-11-15_copy.csv",sep=',')
pd.set_option('display.max_rows', None)
df['Timestamp']=pd.to_datetime(df['Timestamp'])

df['year']=2000
for i in range(0,len(df)):
    df['year'][i]=int(df['Timestamp'][i].year)
print(len(df))
df.head(10)
```

# Crypto

```python

buy_C = df[((df['Transaction Type']== 'buy')|(df['Transaction Type']== 'deposit'))&(df['Asset class']=='Cryptocurrency')&(df['year']==2024)]
sell_C = df[(df['Transaction Type']== 'sell')& (df['Asset class']=='Cryptocurrency')&(df['year']==2024)]
sell_C['Amount Asset']=sell_C['Amount Asset'].astype("float")
buy_C['Restbetrag Asset']=buy_C['Amount Asset'].astype("float")
buy_C['Asset market price']=buy_C['Asset market price'].astype("float")
sell_C['Gewinn']=0.0
sell_C['Gewinn_Steuer']=0.0
sell_C['Euro_In']=0.0
sell_C['Euro_Out']=0.0
#buy_C.head(10) 
#sell_C.head(10)
print(sell_C['Amount Fiat'].sum())
print(buy_C['Amount Fiat'].sum())
#buy_C[(df['Transaction Type']== 'deposit')]
#sell_C
```

```python
taxFree=366
for i in sell_C.index:
    temp = buy_C[(buy_C['Asset']== sell_C['Asset'][i])]
    j=0
    k=temp.index[j]
    
    if(sell_C['Amount Asset'][i]<=buy_C['Restbetrag Asset'][k]):
        buy_C['Restbetrag Asset'][k]-= sell_C['Amount Asset'][i]
        euro_in= sell_C['Amount Asset'][i]*buy_C['Asset market price'][k]
        sell_C['Gewinn'][i]=sell_C['Amount Fiat'][i]- euro_in
        sell_C['Euro_In'][i]=euro_in
        durration=sell_C['Timestamp'][i]-buy_C['Timestamp'][k]
        if(durration.days<taxFree):
            sell_C['Gewinn_Steuer'][i]=sell_C['Gewinn'][i]
            sell_C['Euro_Out'][i]=sell_C['Amount Asset'][i]
        
    else:        
        euro_in=buy_C['Restbetrag Asset'][k]*buy_C['Asset market price'][k]
        rest=sell_C['Amount Asset'][i]-buy_C['Restbetrag Asset'][k]
        buy_C['Restbetrag Asset'][k]=0.0 
        durration=sell_C['Timestamp'][i]-buy_C['Timestamp'][k]
        gewinnSteuerFrei=0.0
        euro_out_free=0.0
        if(durration.days>taxFree):            
            gewinnSteuerFrei=euro_in
            euro_out_free= buy_C['Restbetrag Asset'][k]*sell_C['Asset market price'][i]
            print('test')
            
            
            
        while(rest>0.0000000001):
            j+=1
            k=temp.index[j]   
            if(rest<= buy_C['Restbetrag Asset'][k]):  
                in1=(rest*buy_C['Asset market price'][k])
                euro_in+=in1
                buy_C['Restbetrag Asset'][k]-=rest
                rest_alt=rest
                rest=0.0
                durration=sell_C['Timestamp'][i]-buy_C['Timestamp'][k]
                if(durration.days>taxFree):
                    gewinnSteuerFrei+=in1
                    euro_out_free+= rest_alt*sell_C['Asset market price'][i]
                    print('test')
                
            else:
                in1=buy_C['Restbetrag Asset'][k]*buy_C['Asset market price'][k]
                euro_in+=in1
                rest-=buy_C['Restbetrag Asset'][k]
                buy_C['Restbetrag Asset'][k]=0.0 
                durration=sell_C['Timestamp'][i]-buy_C['Timestamp'][k]
                if(durration.days>taxFree):
                    gewinnSteuerFrei+=in1 
                    euro_out_free+= buy_C['Restbetrag Asset'][k]*sell_C['Asset market price'][i]
                    print('test')
            
            
        sell_C['Gewinn'][i]=sell_C['Amount Fiat'][i]-euro_in  
        sell_C['Gewinn_Steuer'][i]=sell_C['Amount Fiat'][i]-euro_in-gewinnSteuerFrei
        sell_C['Euro_In'][i]=euro_in
        sell_C['Euro_Out'][i]=sell_C['Amount Asset'][i]-euro_out_free
       
        #print(i,sell_C['Asset'][i],sell_C['Amount Asset'][i],temp.index[j],temp['Asset'][temp.index[j]],buy_C['Restbetrag Asset'][temp.index[j]],buy_C['Amount Asset'][temp.index[j]])
        #print(i,sell_C['Asset'][i],sell_C['Amount Asset'][i],k, buy_C['Asset'][k],buy_C['Amount Asset'][k],buy_C['Restbetrag Asset'][k],sell_C['Gewinn'][i],euro_in,sell_C['Amount Fiat'][62])
```
